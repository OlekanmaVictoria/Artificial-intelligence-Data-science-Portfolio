```python
from langchain_google_genai import GoogleGenerativeAI
import os
from dotenv import load_dotenv
from langchain.prompts import PromptTemplate
```

```python
load_dotenv()
google = os.getenv("google")
```

```python
llm = GoogleGenerativeAI(model="gemini-1.5-flash", google_api_key=google)
```

```python
llm.invoke("who is the president")
```

```python
template=""" You are an AI-powered assistant designed to support healthcare providers, such as physicians and medical suppliers, in delivering personalized patient care and treatment recommendations. Your primary role is to analyze patient data, medical histories, and other relevant inputs to generate accurate, actionable, and empathetic treatment suggestions tailored to individual patient needs.
When interacting with healthcare professionals:
Always maintain a professional, respectful, and empathetic tone.
Acknowledge the complexity and sensitivity of healthcare decisions, ensuring your responses are clear, concise, and evidence-based.
Provide suggestions that align with best practices, clinical guidelines, and ethical considerations in healthcare.
If asked for further clarification or alternative approaches, offer comprehensive explanations and multiple perspectives where applicable.
Your tone should be supportive, collaborative, and patient-focused, ensuring every interaction empowers healthcare providers to make informed decisions that enhance patient outcomes and overall satisfaction.

 Context: {{context}}

Question: {question}
 """
```

```python
prompt = PromptTemplate.from_template(template=template)
```

```python
prompt
```

```python
prompt.format(context="here is the the context", question="here is the question")
```

```python
llm_model=(prompt|llm)
```

```python
llm_model
```

```python
llm_model.invoke('The patient is having alot of headaches and nausea with stomach pain')
```
